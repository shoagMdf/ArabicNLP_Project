```python
import os
import pandas as pd
import stanza
from tqdm import trange
import torch
import logging
import json

# -------------------------------
# Logging setup
# -------------------------------
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)

# -------------------------------
# Check GPU
# -------------------------------
device = "cuda" if torch.cuda.is_available() else "cpu"
logger.info(f"Using device: {device}")

# -------------------------------
# SAVE FILES IN GOOGLE DRIVE
# -------------------------------
progress_file = "/content/drive/MyDrive/active_voice_progress.json"
output_partial = "/content/drive/MyDrive/active_voice_partial.csv"

# -------------------------------
# Load or initialize resume progress
# -------------------------------
if os.path.exists(progress_file):
    with open(progress_file, "r") as f:
        progress = json.load(f)
        start_batch_idx = progress.get("last_finished", 0)
else:
    start_batch_idx = 0

logger.info(f"Resuming from batch index: {start_batch_idx}")

# -------------------------------
# Download Arabic Stanza model
# -------------------------------
stanza.download('ar', processors='tokenize,pos,lemma,depparse')

# -------------------------------
# Initialize Stanza pipeline
# -------------------------------
nlp = stanza.Pipeline(
    'ar',
    processors='tokenize,pos,lemma,depparse',
    use_gpu=True if device == "cuda" else False
)

# -------------------------------
# Load dataset
# -------------------------------
df = pd.read_csv('/content/drive/MyDrive/datasets/final_dataset_for_modeling.csv')
df['text'] = df['text'].astype(str).fillna("").str.strip()
df['processed_text'] = df['processed_text'].astype(str).fillna("").str.strip()

# -------------------------------
# Function: Count active voice
# -------------------------------
def count_active_voice_batch(text_list):
    results = []
    for text in text_list:
        doc = nlp(text)
        active = 0
        for sent in doc.sentences:
            deprels = [w.deprel for w in sent.words if w.deprel]
            has_verb = any(w.upos == 'VERB' for w in sent.words)
            if has_verb and any(d.startswith('nsubj') and d != 'nsubj:pass' for d in deprels):
                active += 1
        results.append(active)
    return results

# -------------------------------
# Prepare batch processing
# -------------------------------
batch_size = 256
num_batches = (len(df) + batch_size - 1) // batch_size

active_voice_text = []
active_voice_processed = []

# -------------------------------
# Load partial results if exist
# -------------------------------
if os.path.exists(output_partial):
    saved = pd.read_csv(output_partial)
    active_voice_text = saved["Active_Voice_Count_text"].tolist()
    active_voice_processed = saved["Active_Voice_Count_processed"].tolist()
    logger.info(f"Loaded partial results: {len(active_voice_text)} rows")

    # Fix mismatch between CSV & progress
    expected_rows = start_batch_idx * batch_size
    if len(active_voice_text) < expected_rows:
        logger.warning("Partial results shorter than progress index — fixing.")
        start_batch_idx = len(active_voice_text) // batch_size
    elif len(active_voice_text) > expected_rows:
        logger.warning("Partial results longer than progress index — fixing.")
        start_batch_idx = len(active_voice_text) // batch_size

# -------------------------------
# Processing loop with tqdm + resume
# -------------------------------
pbar = trange(start_batch_idx, num_batches, desc="Processing batches", initial=start_batch_idx)

for batch_idx in pbar:

    start = batch_idx * batch_size
    end = min(start + batch_size, len(df))

    batch_text = df['text'].iloc[start:end].tolist()
    batch_processed = df['processed_text'].iloc[start:end].tolist()

    active_voice_text.extend(count_active_voice_batch(batch_text))
    active_voice_processed.extend(count_active_voice_batch(batch_processed))

    # Save partial after every batch to Drive
    pd.DataFrame({
        "Active_Voice_Count_text": active_voice_text,
        "Active_Voice_Count_processed": active_voice_processed
    }).to_csv(output_partial, index=False)

    # Update progress file in Drive
    with open(progress_file, "w") as f:
        json.dump({"last_finished": batch_idx + 1}, f)

    pbar.set_postfix({"Done": f"{batch_idx+1}/{num_batches}"})

# -------------------------------
# Final save in Drive
# -------------------------------
df['Active_Voice_Count_text'] = active_voice_text
df['Active_Voice_Count_processed'] = active_voice_processed

output_dir = "/content/drive/MyDrive/NLP_outputs"
os.makedirs(output_dir, exist_ok=True)

full_file = os.path.join(output_dir, "active_voice_results_gpu.csv")
summary_file = os.path.join(output_dir, "active_voice_summary_gpu.csv")

df.to_csv(full_file, index=False, encoding="utf-8-sig")

df[['Active_Voice_Count_text', 'Active_Voice_Count_processed']].describe().to_csv(
    summary_file, encoding="utf-8-sig"
)

logger.info(f"✔ Full results saved → {full_file}")
logger.info(f"✔ Summary saved → {summary_file}")
logger.info("🎉 ALL BATCHES COMPLETED SUCCESSFULLY!")
```

```python
import os
import pandas as pd
import stanza
from tqdm import tqdm
import torch
import logging

# -------------------------------
# Logging setup
# -------------------------------
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)

# -------------------------------
# Check GPU
# -------------------------------
device = "cuda" if torch.cuda.is_available() else "cpu"
logger.info(f"Using device: {device}")

# -------------------------------
# Download Arabic Stanza model
# -------------------------------
stanza.download('ar', processors='tokenize,pos,lemma,depparse')

# -------------------------------
# Initialize Stanza pipeline
# -------------------------------
nlp = stanza.Pipeline(
    'ar',
    processors='tokenize,pos,lemma,depparse',
    use_gpu=True if device == "cuda" else False
)

# -------------------------------
# Load dataset from Google Drive
# -------------------------------
df = pd.read_csv('/content/drive/MyDrive/datasets/final_dataset_for_modeling.csv')
df['text'] = df['text'].astype(str).fillna("").str.strip()
df['processed_text'] = df['processed_text'].astype(str).fillna("").str.strip()

# Count active voice function
def count_active_voice_batch(text_list):
    results = []
    for text in text_list:            # process text one-by-one
        doc = nlp(text)               # Stanza requires a string
        active = 0
        for sent in doc.sentences:
            deprels = [w.deprel for w in sent.words if w.deprel]
            has_verb = any(w.upos == 'VERB' for w in sent.words)
            if has_verb and any(d.startswith('nsubj') and d != 'nsubj:pass' for d in deprels):
                active += 1
        results.append(active)
    return results



# GPU batching
batch_size = 256
active_voice_text = []
active_voice_processed = []

for start in tqdm(range(0, len(df), batch_size), desc="Processing batches"):
    end = min(start + batch_size, len(df))
    batch_text = df['text'].iloc[start:end].tolist()
    batch_processed = df['processed_text'].iloc[start:end].tolist()

    active_voice_text.extend(count_active_voice_batch(batch_text))
    active_voice_processed.extend(count_active_voice_batch(batch_processed))

df['Active_Voice_Count_text'] = active_voice_text
df['Active_Voice_Count_processed'] = active_voice_processed

# SAVE RESULTS IN COLAB OR GOOGLE DRIVE

# OPTION 1 — Save inside Colab
output_dir = "/content/output"
os.makedirs(output_dir, exist_ok=True)

# OPTION 2 — Save in Google Drive
# output_dir = "/content/drive/MyDrive/NLP_outputs"
# os.makedirs(output_dir, exist_ok=True)

# Save files
output_full = os.path.join(output_dir, "active_voice_results_gpu.csv")
df[['text', 'processed_text', 'Active_Voice_Count_text', 'Active_Voice_Count_processed']].to_csv(
    output_full, index=False, encoding="utf-8-sig"
)

output_summary = os.path.join(output_dir, "active_voice_summary_gpu.csv")
summary_df = df[['Active_Voice_Count_text', 'Active_Voice_Count_processed']].describe()
summary_df.to_csv(output_summary, index=True, encoding="utf-8-sig")

logger.info(f"Full results saved → {output_full}")
logger.info(f"Summary saved → {output_summary}")

```
